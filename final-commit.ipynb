```python
import subprocess
from pathlib import Path

repo_path = Path('/workspaces/-294612-git-.github.io')

print("=== FINAL COMMIT PREPARATION ===")

# Check current status
result = subprocess.run(['git', 'status', '--short'], cwd=repo_path, capture_output=True, text=True)
print("Current status:")
print(result.stdout)

# Stage all changes
print("\nStaging all changes...")
result = subprocess.run(['git', 'add', '.'], cwd=repo_path, capture_output=True, text=True)
print(f"git add . : rc={result.returncode}")
if result.stderr:
    print("STDERR:", result.stderr)

# Check what's staged
result = subprocess.run(['git', 'diff', '--cached', '--name-only'], cwd=repo_path, capture_output=True, text=True)
staged_files = result.stdout.strip().split('\n') if result.stdout.strip() else []
print(f"\nStaged files ({len(staged_files)}):")
for file in staged_files:
    print(f"  {file}")

# Commit the changes
if staged_files:
    commit_msg = "Fix quiz feedback persistence and clean up Java path resolution"
    print(f"\nCommitting with message: '{commit_msg}'")
    result = subprocess.run(['git', 'commit', '-m', commit_msg], cwd=repo_path, capture_output=True, text=True)
    print(f"git commit : rc={result.returncode}")
    if result.stdout:
        print("STDOUT:", result.stdout)
    if result.stderr:
        print("STDERR:", result.stderr)

    # Final status
    result = subprocess.run(['git', 'status', '--short', '--branch'], cwd=repo_path, capture_output=True, text=True)
    print("\nFinal status:")
    print(result.stdout)
else:
    print("\nNo files to commit!")
```
