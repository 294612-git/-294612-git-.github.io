```python
import subprocess
from pathlib import Path

repo_path = Path('/workspaces/-294612-git-.github.io')

# Clean up temporary files
temp_files = ['check-status.py', 'final-status.ipynb', 'merge-complete.ipynb', 'merge-debug.ipynb']
for file in temp_files:
    file_path = repo_path / file
    if file_path.exists():
        file_path.unlink()
        print(f'Removed {file}')

result = subprocess.run(['git', 'status', '--short', '--branch'], cwd=repo_path, capture_output=True, text=True)
print('CLEAN STATUS:')
print(result.stdout)
```
