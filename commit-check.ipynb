```python
import subprocess
from pathlib import Path

repo_path = Path('/workspaces/-294612-git-.github.io')

# Check branch status
result = subprocess.run(['git', 'status', '--branch', '--short'], cwd=repo_path, capture_output=True, text=True)
print('BRANCH STATUS:')
print(result.stdout)

# Check for uncommitted changes
result = subprocess.run(['git', 'status', '--porcelain'], cwd=repo_path, capture_output=True, text=True)
lines = result.stdout.strip().split('\n')
if lines and lines[0]:
    print('\nFILES WITH CHANGES:')
    for line in lines:
        if line.strip():
            status = line[:2]
            filename = line[2:].strip()
            if status == ' M':
                print(f'  Modified: {filename}')
            elif status == 'A ':
                print(f'  Added: {filename}')
            elif status == ' D':
                print(f'  Deleted: {filename}')
            elif status == '??':
                print(f'  Untracked: {filename}')
            else:
                print(f'  {status}: {filename}')
else:
    print('\nNo changes to commit.')

# Check if there are staged changes
result = subprocess.run(['git', 'diff', '--cached', '--name-only'], cwd=repo_path, capture_output=True, text=True)
staged = result.stdout.strip()
if staged:
    print(f'\nSTAGED FILES ({len(staged.split())}):')
    for file in staged.split('\n'):
        print(f'  {file}')
else:
    print('\nNo staged changes.')
```
