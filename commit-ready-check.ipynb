```python
import subprocess
from pathlib import Path

repo_path = Path('/workspaces/-294612-git-.github.io')

# Get modified files
result = subprocess.run(['git', 'status', '--porcelain'], cwd=repo_path, capture_output=True, text=True)
status_lines = [line for line in result.stdout.strip().split('\n') if line.strip()]

print("=== FILES TO COMMIT ===")
modified_files = []
for line in status_lines:
    if line.strip():
        status = line[:2]
        filename = line[2:].strip()
        if status in [' M', 'A ', 'D ', 'R ', 'C ', 'U ']:
            modified_files.append(filename)
            print(f"  {status} {filename}")

print(f"\nTotal files to commit: {len(modified_files)}")

# Check for basic HTML syntax in key files
print("\n=== QUICK HTML CHECK ===")
key_files = [
    'grade 3/english/grade3-foundations.html',
    'src/main/java/com/example/Main.java',
    'src/main/resources/index2.html',
    'startquiz.html'
]

for file in key_files:
    file_path = repo_path / file
    if file_path.exists():
        try:
            with open(file_path, 'r', encoding='utf-8') as f:
                content = f.read()

            # Basic checks
            checks = []
            if not content.strip().startswith('<!DOCTYPE html>') and file.endswith('.html'):
                checks.append("Missing DOCTYPE")
            if '<html' not in content and file.endswith('.html'):
                checks.append("Missing <html>")
            if '</html>' not in content and file.endswith('.html'):
                checks.append("Missing </html>")

            if checks:
                print(f"❌ {file}: {', '.join(checks)}")
            else:
                print(f"✅ {file}: OK")
        except Exception as e:
            print(f"❌ {file}: Error reading - {e}")
    else:
        print(f"⚠️  {file}: File not found")

print("\n=== NEXT STEPS ===")
if modified_files:
    print("1. Review the changes above")
    print("2. Stage all changes: git add .")
    print("3. Commit with message: git commit -m 'Update quiz files and fixes'")
    print("4. Push to remote: git push")
else:
    print("No files to commit - repository is clean!")
```
